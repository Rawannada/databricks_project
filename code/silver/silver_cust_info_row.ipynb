#init

```python
from pyspark.sql.functions import col,trim
from pyspark.sql.types import StringType
```

#load my df

```python
df = spark.table("dev_project.bronze.cust_info_row")
```

#display and transform
## trimming values

```python
df .display()
```

```python
for field in df.schema.fields :
    if field.dataType == StringType():
        df = df.withColumn(field.name, trim(col(field.name)))
```

```python
df .display()
```

#load into silver 

```python
df .write .mode("overwrite").saveAsTable("dev_project.silver.cust_info")
```
